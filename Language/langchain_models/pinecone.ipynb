```python
from langchain_huggingface import HuggingFaceEmbeddings
from pinecone import Pinecone, ServerlessSpec
```

```python
!pip install sentence-transformers
```

```python
pc=Pinecone(api_key='')
```

```python
from sentence_transformers import SentenceTransformer
emb=SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

```python
indexn='first'
pc.create_index(name=indexn,
               dimension=384,metric='cosine',spec=ServerlessSpec(
                   cloud='aws',region='us-east-1'
               ))
```

```python
from langchain_pinecone import PineconeVectorStore
```

```python
i=pc.Index(indexn)
```

```python
i
```

```python
vec=PineconeVectorStore(index=i,embedding=emb)
```

```python
from langchain_core.documents import Document

docs = [
    Document(page_content="Transformers changed NLP forever"),
    Document(page_content="Machine learning is a subset of AI"),
    Document(page_content="Football is popular worldwide")
]

vec.add_documents(docs)
```

```python
results = vec.similarity_search(
    "Deep learning models in NLP",
    k=2
)

for r in results:
    print(r.page_content)
```
