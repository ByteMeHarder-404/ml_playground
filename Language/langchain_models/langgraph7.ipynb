```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint
from langgraph.checkpoint.memory import InMemorySaver
```

```python
from kaggle_secrets import UserSecretsClient
user_secrets = UserSecretsClient()
secret_value_0 = user_secrets.get_secret("HUGGINGFACEHUB_API_TOKEN")
```

```python
model=HuggingFaceEndpoint(repo_id='Qwen/Qwen3-4B-Instruct-2507',
    temperature=0.1,
    max_new_tokens=2048,
    task='conversational',
    huggingfacehub_api_token=secret_value_0)
llm=ChatHuggingFace(llm=model)
```

```python
class Joker(TypedDict):
    topic:str
    joke:str
    explaination:str
```

```python
def generator(state:Joker):
    p=f'Generate a joke on the topic {state['topic']}'
    res=llm.invoke(p).content
    return {'joke':res}
```

```python
def explaination(state:Joker):
    p=f'Write an human explaination for the joke -{state['joke']}'
    res=llm.invoke(p).content
    return {'explaination':res}
```

```python
graph=StateGraph(Joker)
```

```python
graph.add_node('generate_joke', generator)
graph.add_node('generate_explanation', explaination)
```

```python
graph.add_edge(START, 'generate_joke')
graph.add_edge('generate_joke', 'generate_explanation')
graph.add_edge('generate_explanation', END)
```

```python
chk=InMemorySaver()
```

```python
a=graph.compile(checkpointer=chk)
```

```python
c1={'configurable':{'thread_id':'1'}}
```

```python
a.invoke({'topic':'Iran'},config=c1)
```
