```python
import random
```

```python
def draw():
    val=random.randint(1,10)
    if random.random()<(2/3):
        return val
    else:
        return -val
```

```python
class E21:
    def __init__(self):
        self.deal=None
        self.play=None
        self.term=False
    def reset(self):
        self.deal=random.randint(1,10)
        self.play=random.randint(1,10)
        self.term=False
        return (self.deal,self.play)
    def step(self,act):
        if self.term:
            raise Exception('Game over....reset()')
        if act=='hit':
            self.play+=draw()
            if self.play<1 or self.play>21:
                self.term=True
                return None,-1,True
            return (self.deal,self.play),0,False
        elif act=='stick':
            deals=self.deal
            while 1<=deals<17:
                deals+=draw()
            self.term=True
            if deals<1 or deals>21:
                return None,+1,True
            if self.play>deals:
                return None,+1,True
            elif self.play<deals:
                return None,-1,True
            else:
                return None,0,True
        else:
            raise ValueError('INvalid action')
```

```python
env=E21(
)
```

```python
state=env.reset()
```

```python
done = False

while not done:
    action = random.choice(['hit', 'stick'])
    next_state, reward, done = env.step(action)

    print(f"State: {state}, Action: {action}")
    print(f"Next: {next_state}, Reward: {reward}, Done: {done}")
    print("------")

    state = next_state
```
